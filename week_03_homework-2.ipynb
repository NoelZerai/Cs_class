9. (3 pts) On average are Japanese cars more fuel-efficient than US cars? Write a function that can handle this comparison.  

```python
print(df.columns)
```

```python
is_Japan = df['Origin'] == 'Japan'
is_us = df['Origin'] == 'US'

average_fuel_efficiency_Japan = df.loc[is_Japan, 'MPG'].mean()
average_fuel_efficiency_US = df.loc[is_us, 'MPG'].mean()

print("Average fuel efficiency for Japanese cars: ", average_fuel_efficiency_Japan)
print("Average fuel efficiency for US cars:", average_fuel_efficiency_US)

if average_fuel_efficiency_Japan > average_fuel_efficiency_US:
    print("Japanese cars are more fuel efficient on average,")
elif average_fuel_efficiency_Japan < average_fuel_efficiency_US:
    print("US cars are more fuel efficient on average")
else:
    print("Japanese cars and US cars have the same average fuel efficiency,")
```

11. (2 pts) Create a new pandas dataframe that has the average mpg, weight, and horsepower per country of origin

```python
average_stats_per_country = df.groupby('Origin').agg({
    'MPG': 'mean',
    'Weight': 'mean',
    'Horsepower': 'mean'
})
print(average_stats_per_country)
```
